```python
# ============================================================
# SUMMARY STATISTICS FOR autoparts.csv
# Designed to run directly in Google Colab
# ============================================================

# Import pandas, the main Python library for working with tabular data
import pandas as pd


# ------------------------------------------------------------
# 1. Load the dataset
# ------------------------------------------------------------

# Google Colab stores uploaded files in the "/content" folder.
# This assumes that the uploaded file is named "autoparts.csv".
file_path = "/content/autoparts.csv"

# Read the CSV file and store it as a pandas DataFrame
df = pd.read_csv(file_path)


# ------------------------------------------------------------
# 2. Inspect the dataset
# ------------------------------------------------------------

# Display the first five rows to verify that the file loaded correctly
print("First five rows of the dataset:")
display(df.head())

# Display the number of observations and variables
print(f"\nNumber of observations: {df.shape[0]}")
print(f"Number of variables: {df.shape[1]}")

# Display the variable names and their data types
print("\nVariable names and data types:")
print(df.dtypes)


# ------------------------------------------------------------
# 3. Select numeric variables
# ------------------------------------------------------------

# Summary statistics such as the mean and standard deviation
# are normally calculated for numeric variables.
numeric_df = df.select_dtypes(include="number")

print("\nNumeric variables included in the analysis:")
print(numeric_df.columns.tolist())


# ------------------------------------------------------------
# 4. Compute summary statistics
# ------------------------------------------------------------

# Calculate:
# - Count: number of non-missing observations
# - Mean: arithmetic average
# - Standard deviation: sample standard deviation
# - Minimum: smallest observed value
# - 25%: first quartile
# - Median: middle value, also called the 50th percentile
# - 75%: third quartile
# - Maximum: largest observed value
summary_statistics = numeric_df.describe().T

# Rename the columns to make the output easier to understand
summary_statistics = summary_statistics.rename(
    columns={
        "count": "Count",
        "mean": "Mean",
        "std": "Standard deviation",
        "min": "Minimum",
        "25%": "25th percentile",
        "50%": "Median",
        "75%": "75th percentile",
        "max": "Maximum"
    }
)

# Add the number of missing observations for each variable
summary_statistics["Missing values"] = numeric_df.isna().sum()

# Arrange the columns in a logical order
summary_statistics = summary_statistics[
    [
        "Count",
        "Missing values",
        "Mean",
        "Standard deviation",
        "Minimum",
        "25th percentile",
        "Median",
        "75th percentile",
        "Maximum"
    ]
]

# Give the index a descriptive name
summary_statistics.index.name = "Variable"


# ------------------------------------------------------------
# 5. Display the results
# ------------------------------------------------------------

print("\nSummary statistics:")

# Display the table rounded to two decimal places.
# Rounding affects only how the results appear, not the calculations.
display(summary_statistics.round(2))


# ------------------------------------------------------------
# 6. Optional: save the results as a CSV file
# ------------------------------------------------------------

# Remove the "#" symbols below if you want to save the results.
# The output file will appear in the Colab "/content" folder.

# output_path = "/content/summary_statistics.csv"
# summary_statistics.to_csv(output_path)
# print(f"Summary statistics saved to: {output_path}")
```
